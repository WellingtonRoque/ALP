# 🧠 Parte 2 — Introdução aos Menus

# 📖 O que é um menu?

Menus permitem navegar entre funcionalidades do sistema.



# 💡 Exemplos reais

✅ caixas eletrônicos  
✅ sistemas bancários  
✅ sistemas comerciais  
✅ aplicativos  
✅ plataformas web  



# 💻 Exemplo simples de menu


```python

print("1 - Cadastrar")
print("2 - Listar")
print("0 - Sair")
```


# 🔍 O problema do exemplo

O sistema apenas mostra opções.

Ele ainda não possui interação.



# 🧠 Parte 3 — Menus Interativos

# 💻 Menu utilizando WHILE


```python

opcao = ""

while opcao != "0":

    print("===== SISTEMA =====")
    print("1 - Cadastrar")
    print("2 - Listar")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

print("Sistema encerrado")
```


# 🔍 O que aconteceu?

Enquanto a opção for diferente de 0:

✅ o menu continua executando



# 💻 Menu com decisões


```python

opcao = ""

while opcao != "0":

    print("1 - Produtos")
    print("2 - Clientes")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":
        print("Área de produtos")

    if opcao == "2":
        print("Área de clientes")

print("Sistema finalizado")
```


# 💻 Sistema de login simples


```python

senha = ""

while senha != "1234":

    senha = input("Digite a senha: ")

print("Acesso liberado")
```


# 🧠 Parte 4 — Navegação Computacional

# 📖 Sistemas utilizam navegação constantemente



# 💡 Exemplos reais

✅ menus administrativos  
✅ plataformas online  
✅ ERPs  
✅ caixas eletrônicos  
✅ sistemas de estoque  



# 💻 Navegação entre áreas


```python

opcao = ""

while opcao != "0":

    print("===== MENU =====")
    print("1 - Produtos")
    print("2 - Estoque")
    print("3 - Relatórios")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":
        print("Abrindo produtos")

    if opcao == "2":
        print("Abrindo estoque")

    if opcao == "3":
        print("Abrindo relatórios")

print("Sistema encerrado")
```

# 💻 Exemplo — Caixa eletrônico


```python

saldo = 1000

opcao = ""

while opcao != "0":

    print("===== CAIXA =====")
    print("1 - Ver saldo")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":
        print("Saldo:", saldo)

print("Sistema encerrado")
```


# 💻 Exemplo — Cadastro simples


```python

opcao = ""

while opcao != "0":

    print("1 - Cadastrar produto")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":

        produto = input("Digite o produto: ")

        print("Produto cadastrado:", produto)
```
