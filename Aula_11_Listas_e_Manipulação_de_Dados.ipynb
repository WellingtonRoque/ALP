# 🧠 Parte 2 — Introdução às Listas

# 📖 O que é uma lista?

Lista é uma estrutura utilizada para armazenar múltiplos valores.



# 💻 Exemplo simples


````python

produtos = ["Mouse", "Teclado", "Monitor"]

print(produtos)
````


# 🔍 O que aconteceu?

A variável armazenou vários valores simultaneamente.



# 📖 Estrutura de uma lista

```python
lista = [valor1, valor2, valor3]
```



# 💻 Acessando posições


````python

produtos = ["Mouse", "Teclado", "Monitor"]

print(produtos[0])

print(produtos[1])

print(produtos[2])
````


# ⚠️ Importante

As listas iniciam na posição:

```python
0
```



# 💻 Exemplo visual

| Índice | Valor |
|---|---|
| 0 | Mouse |
| 1 | Teclado |
| 2 | Monitor |



# 🧠 Parte 3 — Manipulação de Listas

# 📖 Adicionando elementos


````python

produtos = []

produtos.append("Notebook")

produtos.append("Mouse")

print(produtos)
````


# 🔍 append()

Adiciona novos elementos na lista.



# 📖 Removendo elementos


````python

produtos = ["Mouse", "Teclado", "Monitor"]

produtos.remove("Teclado")

print(produtos)
````


# 📖 Alterando elementos


````python

produtos = ["Mouse", "Teclado"]

produtos[1] = "Monitor"

print(produtos)
````


# 📖 Tamanho da lista


````python

produtos = ["Mouse", "Teclado", "Monitor"]

print(len(produtos))
````


# 🔍 len()

Retorna quantidade de elementos.



# 🧠 Parte 4 — Percorrendo Listas

# 💻 Utilizando FOR


````python

produtos = ["Mouse", "Teclado", "Monitor"]

for produto in produtos:

    print(produto)
````


# 💻 Utilizando índices


````python

produtos = ["Mouse", "Teclado", "Monitor"]

for indice in range(len(produtos)):

    print(indice, produtos[indice])
````


# 💻 Soma de valores


````python

valores = [10, 20, 30, 40]

total = 0

for valor in valores:

    total = total + valor

print(total)
````


# 🧠 Parte 5 — Listas em Sistemas Reais

# 💡 Sistemas utilizam listas constantemente



# 💡 Exemplos reais

✅ produtos  
✅ clientes  
✅ estoque  
✅ relatórios  
✅ pedidos  



# 💻 Exemplo — Cadastro de produtos


````python

produtos = []

for contador in range(3):

    produto = input("Digite o produto: ")

    produtos.append(produto)

print(produtos)
````


# 💻 Exemplo — Sistema de estoque


````python

estoque = [10, 5, 8]

for quantidade in estoque:

    print("Quantidade:", quantidade)
````


# 💻 Exemplo — Relatório de clientes


````python

clientes = ["Carlos", "Maria", "João"]

for cliente in clientes:

    print(cliente)
````


# 🛒 Parte 6 — Projeto do Semestre

# 📦 Sistema de Gestão Comercial

Hoje iremos adicionar armazenamento de dados utilizando listas.



# 💻 Cadastro de produtos


````python

produtos = []

opcao = ""

while opcao != "0":

    print("1 - Cadastrar produto")
    print("2 - Listar produtos")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":

        produto = input("Digite o produto: ")

        produtos.append(produto)

        print("Produto cadastrado")

    if opcao == "2":

        print(produtos)
````


# 💻 Sistema de clientes


````python

clientes = []

for contador in range(3):

    cliente = input("Digite o nome do cliente: ")

    clientes.append(cliente)

print(clientes)
````
