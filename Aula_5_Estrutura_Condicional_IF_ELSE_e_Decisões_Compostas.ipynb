# 🧠 Parte 2 — Introdução ao IF ELSE

# 📖 O que é IF ELSE?

O IF ELSE permite criar dois caminhos possíveis no programa.



# 💻 Exemplo simples

```python
idade = 20

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
```


````python

idade = 20

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
````


# 🔍 O que aconteceu?

Se a condição for verdadeira:

✅ executa IF

Caso contrário:

✅ executa ELSE



# 📖 Estrutura do IF ELSE

```python
if condição:
    bloco verdadeiro
else:
    bloco falso
```



# ⚠️ Importância da indentação

Python utiliza indentação para organizar os blocos.



# 💻 Exemplo — Sistema de login


````python

senha = input("Digite a senha: ")

if senha == "1234":
    print("Acesso liberado")
else:
    print("Senha incorreta")
````


# 💻 Exemplo — Aprovação de aluno


````python

media = float(input("Digite a média: "))

if media >= 6:
    print("Aluno aprovado")
else:
    print("Aluno reprovado")
````


# 🧠 Parte 4 — Fluxo de Execução

# 📖 Como o IF ELSE funciona?

O Python:

1. verifica condição  
2. executa IF se verdadeiro  
3. executa ELSE se falso  



# 💻 Fluxo visual

```plaintext
Condição → True → Executa IF
Condição → False → Executa ELSE
```



# 💻 Exemplo


````python

idade = 15

if idade >= 18:
    print("Pode dirigir")
else:
    print("Não pode dirigir")
````
