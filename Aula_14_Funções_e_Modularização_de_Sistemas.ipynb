# 🧠 Parte 2 — Introdução às Funções

# 📖 O que é uma função?

Funções são blocos reutilizáveis de código.



# 💡 Vantagens das funções

✅ organização  
✅ reutilização  
✅ manutenção  
✅ legibilidade  



# 💻 Exemplo simples


````python

def mensagem():

    print("Olá Mundo")

mensagem()
````


# 🔍 Explicação

```python
def
```

Define uma função.



# 📖 Estrutura básica

```python

    bloco
```



# 💻 Exemplo — Saudação


````python

def saudacao():

    print("Bem-vindo ao sistema")

saudacao()
````


# 💻 Exemplo — Soma simples


````python

def somar():

    numero1 = 10

    numero2 = 20

    resultado = numero1 + numero2

    print(resultado)

somar()
````


# 🧠 Parte 3 — Parâmetros e Argumentos

# 📖 O que são parâmetros?

Permitem enviar dados para funções.



# 💻 Exemplo


````python

def saudacao(nome):

    print("Olá", nome)

saudacao("Carlos")
````


# 🔍 Explicação

```python
nome
```

é um parâmetro.



# 💻 Soma utilizando parâmetros


````python

def somar(numero1, numero2):

    resultado = numero1 + numero2

    print(resultado)

somar(10, 20)
````


# 💻 Produto utilizando parâmetros


````python

def cadastrar_produto(produto, preco):

    print("Produto:", produto)

    print("Preço:", preco)

cadastrar_produto("Notebook", 3500)
````


# 🧠 Parte 4 — Retorno de Valores

# 📖 O que é retorno?

Permite devolver valores da função.



# 💻 Exemplo com return


````python

def somar(numero1, numero2):

    resultado = numero1 + numero2

    return resultado

valor = somar(10, 20)

print(valor)
````


# 💻 Exemplo — Média


````python

def calcular_media(nota1, nota2):

    media = (nota1 + nota2) / 2

    return media

resultado = calcular_media(8, 6)

print(resultado)
````


# 💻 Exemplo — Desconto


````python

def calcular_desconto(valor):

    desconto = valor * 0.10

    return desconto

resultado = calcular_desconto(1000)

print(resultado)
````


# 🧠 Parte 5 — Modularização de Sistemas

# 📖 O que é modularização?

Dividir o sistema em pequenas partes organizadas.



# 💡 Sistemas reais utilizam modularização

✅ ERPs  
✅ sistemas bancários  
✅ plataformas web  
✅ jogos  
✅ aplicativos  



# 💻 Sistema sem função


````python

produto = input("Digite o produto: ")

preco = float(input("Digite o preço: "))

print(produto)

print(preco)
````


# 💻 Sistema modularizado


````python

def cadastrar_produto():

    produto = input("Digite o produto: ")

    preco = float(input("Digite o preço: "))

    print(produto)

    print(preco)

cadastrar_produto()
````


# 💻 Sistema com múltiplas funções


````python

def menu():

    print("1 - Produtos")
    print("0 - Sair")

def cadastrar():

    produto = input("Digite o produto: ")

    print("Produto cadastrado")

menu()

cadastrar()
````


# 🛒 Parte 6 — Projeto do Semestre

# 📦 Sistema de Gestão Comercial

Hoje iremos modularizar o sistema utilizando funções.



# 💻 Sistema modularizado


````python

produtos = []

def menu():

    print("===== SISTEMA =====")
    print("1 - Cadastrar produto")
    print("2 - Listar produtos")
    print("0 - Sair")

def cadastrar():

    produto = input("Digite o produto: ")

    produtos.append(produto)

    print("Produto cadastrado")

def listar():

    print(produtos)

opcao = ""

while opcao != "0":

    menu()

    opcao = input("Digite a opção: ")

    if opcao == "1":

        cadastrar()

    if opcao == "2":

        listar()
````
