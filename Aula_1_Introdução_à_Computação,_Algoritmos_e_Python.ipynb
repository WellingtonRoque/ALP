# 🧠 Parte 2 — Introdução à Computação

# 💻 O que é um computador?

Um computador é uma máquina capaz de:

✅ receber dados  
✅ processar informações  
✅ armazenar informações  
✅ gerar resultados  



# 🧱 Componentes básicos

## Processador (CPU)

Responsável pelos cálculos.

---

## Memória RAM

Armazena dados temporariamente.

---

## Armazenamento

HD ou SSD.

---

## Entrada

Teclado, mouse etc.

---

## Saída

Monitor, impressora etc.



# ⚙️ Como um programa funciona?

Todo software segue:

```plaintext
Entrada → Processamento → Saída
```

# 🧠 Parte 3 — O que é Algoritmo?

# 📖 Definição

Algoritmo é:

# 🔥 Uma sequência lógica de passos para resolver um problema.



# 💡 Exemplo do mundo real

## Fazer café

1. pegar água  
2. colocar café  
3. ferver água  
4. passar café  
5. servir  

👉 isso é um algoritmo.



# 💻 Exemplo computacional

## Somar dois números

1. receber número 1  
2. receber número 2  
3. somar  
4. mostrar resultado  



# ⚠️ Diferença entre Algoritmo e Programa

## Algoritmo

👉 solução lógica.

## Programa

👉 implementação em linguagem de programação.



# 🧠 Parte 4 — Lógica de Programação

# 📖 O que é lógica?

É a capacidade de resolver problemas de forma organizada.



# 💡 Exemplo

Problema:

```plaintext
Calcular média de notas
```

---

# 🔍 Pensamento lógico

Passos:

1. receber nota 1  
2. receber nota 2  
3. somar  
4. dividir por 2  
5. mostrar resultado  



# 🧱 Estruturas fundamentais

Todo algoritmo utiliza:

✅ repetição
