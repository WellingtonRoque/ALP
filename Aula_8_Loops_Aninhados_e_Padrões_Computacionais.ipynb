# 🧠 Parte 2 — Introdução aos Loops Aninhados

# 📖 O que são loops aninhados?

Loops aninhados são estruturas de repetição dentro de outras estruturas de repetição.



# 💡 Exemplo visual

```plaintext
FOR externo
    FOR interno
```



# 💻 Exemplo simples


````python

for linha in range(3):

    for coluna in range(3):

        print(coluna)
````


# 🔍 O que aconteceu?

O FOR interno executa completamente para cada execução do FOR externo.



# 🧠 Parte 3 — FOR dentro de FOR

# 💻 Exemplo — Tabela numérica


````python

for linha in range(3):

    for coluna in range(3):

        print(linha, coluna)
````


# 💻 Exemplo — Tabuadas


````python

for numero in range(1, 6):

    print("Tabuada do", numero)

    for contador in range(1, 11):

        resultado = numero * contador

        print(resultado)
````


# 💻 Exemplo — Contagem bidimensional


````python

for linha in range(5):

    for coluna in range(5):

        print("*", end=" ")

    print()
````


# 🔍 Explicação do end

```python
end=" "
```

Evita quebra automática de linha.



# 🧠 Parte 4 — Padrões Computacionais

# 📖 O que são padrões?

São estruturas repetitivas organizadas.



# 💻 Quadrado de caracteres


````python

for linha in range(5):

    for coluna in range(5):

        print("#", end=" ")

    print()
````


# 💻 Triângulo de caracteres


````python

for linha in range(1, 6):

    print("*" * linha)
````


# 💻 Sequência numérica


````python

for linha in range(1, 6):

    for coluna in range(1, 6):

        print(coluna, end=" ")

    print()
````


# 💻 Multiplicação em tabela


````python

for linha in range(1, 6):

    for coluna in range(1, 6):

        resultado = linha * coluna

        print(resultado, end=" ")

    print()
````

# 💻 Exemplo — Relatório de vendas


````python

for semana in range(1, 5):

    print("Semana", semana)

    for venda in range(1, 4):

        print("Venda", venda)
````


# 💻 Exemplo — Controle de estoque


````python

for corredor in range(1, 4):

    print("Corredor", corredor)

    for produto in range(1, 6):

        print("Produto", produto)
````
