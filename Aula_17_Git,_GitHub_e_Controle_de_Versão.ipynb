# 🧠 Parte 2 — Introdução ao Versionamento

# 📖 O que é versionamento?

Versionamento é o controle das alterações realizadas em um projeto.



# 💡 Problemas sem versionamento

❌ perda de código  
❌ sobrescrita de arquivos  
❌ dificuldade de trabalho em equipe  
❌ ausência de histórico  



# 💡 Benefícios do versionamento

✅ histórico de alterações  
✅ recuperação de versões  
✅ trabalho colaborativo  
✅ segurança do projeto  



# 📖 O que é Git?

Git é um sistema de controle de versão distribuído.



# 📖 O que é GitHub?

GitHub é uma plataforma para hospedagem de repositórios Git.



# 💡 Empresas utilizam Git constantemente

✅ desenvolvimento web  
✅ inteligência artificial  
✅ automação  
✅ sistemas corporativos  



# 🧠 Parte 3 — Git e Comandos Básicos

# 💻 Inicializando repositório


````python

git init
````


# 🔍 O que aconteceu?

O Git criou uma estrutura de versionamento no projeto.



# 💻 Verificando status


````python

git status
````


# 💻 Adicionando arquivos


````python

git add .
````


# 💻 Criando commit


````python

git commit -m "Primeiro commit"
````


# 💻 Visualizando histórico


````python

git log
````


# 💻 Configurando usuário


````python

git config --global user.name "Seu Nome"

git config --global user.email "email@email.com"
````


# 🧠 Parte 4 — GitHub e Repositórios Remotos

# 📖 Criando repositório no GitHub

Passos:

1. acessar GitHub  
2. criar novo repositório  
3. definir nome  
4. criar repositório  



# 💻 Conectando projeto ao GitHub


````python

git remote add origin URL_DO_REPOSITORIO
````


# 💻 Enviando projeto


````python

git push -u origin main
````


# 💻 Clonando projeto


````python

git clone URL_DO_REPOSITORIO
````


# 💻 Atualizando projeto


````python

git pull
````


# 🧠 Parte 5 — Fluxo Profissional de Desenvolvimento
