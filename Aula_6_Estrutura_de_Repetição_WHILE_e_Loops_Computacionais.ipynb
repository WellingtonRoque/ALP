# 🧠 Parte 2 — Introdução às Estruturas de Repetição

# 📖 O que é repetição?

Estruturas de repetição permitem executar um bloco várias vezes.



# 💡 Exemplo do cotidiano

## Situação

Enquanto houver clientes:

👉 atender cliente



# 💻 Exemplo computacional

```python
contador = 1

while contador <= 5:
    print(contador)

    contador = contador + 1
```


````python

contador = 1

while contador <= 5:
    print(contador)

    contador = contador + 1
````


# 🔍 O que aconteceu?

O programa:

1. iniciou contador  
2. verificou condição  
3. executou bloco  
4. incrementou contador  
5. repetiu processo  



# 🧠 Parte 3 — Estrutura WHILE

# 📖 Sintaxe do WHILE

```python
while condição:
    bloco de código
```



# ⚠️ Importância da atualização da variável

Sem atualizar a variável:

❌ loop infinito



# ❌ Exemplo incorreto

```python
contador = 1

while contador <= 5:
    print(contador)
```



# 💻 Exemplo — Contagem simples


````python

contador = 1

while contador <= 10:
    print(contador)

    contador = contador + 1
````


# 💻 Exemplo — Contagem regressiva


````python

contador = 10

while contador >= 1:
    print(contador)

    contador = contador - 1
````


# 💻 Exemplo — Soma acumulativa


````python

contador = 1

soma = 0

while contador <= 5:

    soma = soma + contador

    contador = contador + 1

print(soma)
````


# 🧠 Parte 4 — Contadores e Acumuladores

# 📖 Contador

Utilizado para controlar repetições.



# 📖 Acumulador

Utilizado para somar valores progressivamente.



# 💻 Exemplo — Contador


````python

contador = 1

while contador <= 3:

    print("Executando")

    contador = contador + 1
````


# 💻 Exemplo — Acumulador


````python

contador = 1

total = 0

while contador <= 5:

    total = total + contador

    contador = contador + 1

print(total)
````


# 🧠 Parte 5 — Loops em Sistemas Reais

# 💡 Sistemas utilizam repetição constantemente



# 💡 Exemplos reais

✅ caixas eletrônicos  
✅ sistemas de login  
✅ menus  
✅ estoque  
✅ processamento de dados  



# 💻 Exemplo — Sistema de menu


````python

opcao = ""

while opcao != "0":

    print("1 - Cadastrar")
    print("2 - Listar")
    print("0 - Sair")

    opcao = input("Digite a opção: ")
````


# 💻 Exemplo — Sistema de estoque


````python

estoque = 5

while estoque > 0:

    print("Produto vendido")

    estoque = estoque - 1

print("Estoque finalizado")
````


# 🧠 Parte 6 — Projeto do Semestre

# 📦 Sistema de Gestão Comercial

Hoje iremos adicionar repetição ao sistema.



# 💻 Sistema de menu inicial


````python

opcao = ""

while opcao != "0":

    print("===== SISTEMA =====")
    print("1 - Produtos")
    print("2 - Clientes")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

print("Sistema encerrado")
````


# 💻 Sistema de vendas


````python

contador = 1

while contador <= 3:

    produto = input("Digite o produto: ")

    print("Produto cadastrado:", produto)

    contador = contador + 1
````
