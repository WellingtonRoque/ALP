# 🧠 Parte 2 — Variáveis

# 📖 O que é uma variável?

Variáveis são espaços na memória utilizados para armazenar informações.



# 💡 Exemplo

```python
nome = "Carlos"
idade = 20
```


````python

nome = "Carlos"

idade = 20

print(nome)

print(idade)
````


# 🔍 Explicação

## nome
Armazena texto.

## idade
Armazena número inteiro.



# ⚠️ Regras para criação de variáveis

✅ não iniciar com números  
✅ não utilizar espaços  
✅ utilizar nomes claros  



# 💡 Exemplos corretos

```python
nome_cliente = "Maria"

valor_produto = 150
```



# ❌ Exemplos incorretos

```python
1nome = "Carlos"

valor total = 100
```


# ⚠️ Problema comum

O input sempre retorna texto.



# ❌ Exemplo incorreto

```python
numero1 = input("Digite um número: ")

numero2 = input("Digite outro número: ")

resultado = numero1 + numero2
```


````python

numero1 = input("Digite um número: ")

numero2 = input("Digite outro número: ")

resultado = numero1 + numero2

print(resultado)
````


# 🔍 O que aconteceu?

O Python concatenou os textos.



# ✅ Forma correta

```python
numero1 = int(input("Digite um número: "))

numero2 = int(input("Digite outro número: "))

resultado = numero1 + numero2
```


````python

numero1 = int(input("Digite um número: "))

numero2 = int(input("Digite outro número: "))

resultado = numero1 + numero2

print(resultado)
````


# 🧠 Parte 5 — Operadores Matemáticos

# 📖 Principais operadores



| Operador | Função |
|---|---|
| + | Soma |
| - | Subtração |
| * | Multiplicação |
| / | Divisão |
| // | Divisão inteira |
| % | Resto da divisão |
| ** | Potência |



# 💻 Soma


````python

numero1 = 10

numero2 = 5

resultado = numero1 + numero2

print(resultado)
````


# 💻 Subtração


````python

resultado = 20 - 5

print(resultado)
````


# 💻 Multiplicação


````python

resultado = 10 * 3

print(resultado)
````


# 💻 Divisão


````python

resultado = 10 / 2

print(resultado)
````


# 💻 Potência


````python

resultado = 2 ** 3

print(resultado)
````


# 💻 Resto da divisão


````python

resultado = 10 % 3

print(resultado)
````

# 🔹 Exercício Guiado 1

Criar programa que:

✅ receba nome do produto  
✅ receba preço  
✅ receba quantidade  
✅ calcule valor total  


````python

produto = input("Digite o produto: ")

preco = float(input("Digite o preço: "))

quantidade = int(input("Digite a quantidade: "))

total = preco * quantidade

print("Produto:", produto)

print("Valor Total:", total)
````


# 🔹 Exercício Guiado 2

Criar sistema que calcule média de notas.


````python

nota1 = float(input("Digite a primeira nota: "))

nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print("Média:", media)
````
