# 🧠 Parte 2 — Introdução às Estruturas Condicionais

# 📖 O que é uma estrutura condicional?

Estruturas condicionais permitem que o programa tome decisões.



# 💡 Exemplo do cotidiano

## Situação

Se estiver chovendo:

👉 levar guarda-chuva

Se não estiver:

👉 sair normalmente



# 💻 Exemplo computacional

```python
idade = 20

if idade >= 18:
    print("Maior de idade")
```


````python

idade = 20

if idade >= 18:
    print("Maior de idade")
````


# 🔍 O que aconteceu?

O programa verificou a condição:

```python
idade >= 18
```

Como o resultado foi verdadeiro:

```python
True
```

o bloco foi executado.



# 🧠 Parte 3 — Estrutura IF

# 📖 Sintaxe do IF

```python
if condição:
    bloco de código
```



# ⚠️ Importância da indentação

Python utiliza indentação para identificar blocos.



# ✅ Exemplo correto

```python
if True:
    print("Executando")
```


````python

if True:
    print("Executando")
````


# ❌ Exemplo incorreto

```python
if True:
print("Erro")
```



# 💻 Exemplo — Validação de estoque


````python

estoque = 10

if estoque > 0:
    print("Produto disponível")
````


# 💻 Exemplo — Validação de senha


````python

senha = input("Digite a senha: ")

if senha == "1234":
    print("Acesso liberado")
````


# 💻 Exemplo — Produto premium


````python

preco = float(input("Digite o preço: "))

if preco >= 1000:
    print("Produto premium")
````

# 💻 Exemplo — Sistema de login


# 💻 Exemplo — Aprovação de aluno


````python

media = float(input("Digite a média: "))

if media >= 6:
    print("Aluno aprovado")
````


# 💻 Exemplo — Controle de estoque


````python

estoque = int(input("Digite o estoque: "))

if estoque <= 5:
    print("Estoque baixo")
````


# 🧠 Parte 5 — Fluxo de Execução

# 📖 Como o IF funciona?

O Python executa:

1. verifica condição  
2. avalia verdadeiro ou falso  
3. executa bloco se verdadeiro  



# 💻 Fluxo visual

```plaintext
Condição → True → Executa bloco
Condição → False → Ignora bloco
```



# 💻 Exemplo


````python

idade = 15

if idade >= 18:
    print("Pode dirigir")

print("Fim do sistema")
````


# 🔍 O que aconteceu?

A condição foi falsa.

Então o Python ignorou o bloco.
