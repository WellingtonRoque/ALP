
# 🧠 Parte 3 — Tratamento de Exceções

# 📖 O que é tratamento de erros?

Permite evitar que o sistema seja encerrado inesperadamente.



# 📖 Estrutura básica

```python
try:
    bloco

except:
    tratamento
```



# 💻 Exemplo simples


````python

try:

    numero = int(input("Digite um número: "))

    print(numero)

except:

    print("Valor inválido")
````


# 🔍 O que aconteceu?

Se o usuário digitar valor incorreto:

✅ o sistema continua funcionando



# 💻 Exemplo — Divisão segura


````python

try:

    numero1 = int(input("Digite um número: "))

    numero2 = int(input("Digite outro número: "))

    resultado = numero1 / numero2

    print(resultado)

except:

    print("Erro na operação")
````


# 💻 Capturando tipos específicos


````python

try:

    numero = int(input("Digite um número: "))

    print(numero)

except ValueError:

    print("Digite apenas números")
````


# 💻 Tratando divisão por zero


````python

try:

    numero1 = int(input("Digite um número: "))

    numero2 = int(input("Digite outro número: "))

    resultado = numero1 / numero2

    print(resultado)

except ZeroDivisionError:

    print("Não é possível dividir por zero")
````


# 🧠 Parte 4 — Validação de Dados

# 📖 O que é validação?

Processo de verificar se os dados estão corretos.



# 💡 Sistemas precisam validar dados constantemente

✅ login  
✅ cadastro  
✅ pagamentos  
✅ notas  
✅ estoque  



# 💻 Exemplo — Validação de idade


````python

idade = int(input("Digite sua idade: "))

if idade >= 18:

    print("Maior de idade")

else:

    print("Menor de idade")
````


# 💻 Exemplo — Validação de preço


````python

preco = float(input("Digite o preço: "))

if preco > 0:

    print("Preço válido")

else:

    print("Preço inválido")
````


# 💻 Exemplo — Login seguro


````python

usuario = input("Digite o usuário: ")

senha = input("Digite a senha: ")

if usuario == "admin" and senha == "1234":

    print("Acesso liberado")

else:

    print("Login inválido")
````

# 💻 Exemplo — Cadastro seguro


````python

try:

    nome = input("Digite o nome: ")

    idade = int(input("Digite a idade: "))

    print(nome)

    print(idade)

except ValueError:

    print("Idade inválida")
````


# 💻 Exemplo — Arquivo seguro


````python

try:

    arquivo = open("clientes.txt", "r")

    print(arquivo.read())

    arquivo.close()

except:

    print("Arquivo não encontrado")
````

# 💻 Cadastro seguro de produtos


````python

def cadastrar_produto():

    try:

        produto = input("Digite o produto: ")

        preco = float(input("Digite o preço: "))

        arquivo = open("produtos.txt", "a")

        arquivo.write(produto + " - " + str(preco) + "\n")

        arquivo.close()

        print("Produto cadastrado")

    except ValueError:

        print("Preço inválido")

cadastrar_produto()
````
