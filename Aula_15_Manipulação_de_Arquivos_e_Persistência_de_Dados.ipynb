# 🧠 Parte 2 — Introdução aos Arquivos

# 📖 O que é um arquivo?

Arquivo é uma estrutura utilizada para armazenar dados permanentemente.



# 💡 Exemplos reais

✅ sistemas comerciais  
✅ bancos  
✅ planilhas  
✅ relatórios  
✅ cadastros  



# 📖 Principais modos de abertura

| Modo | Função |
|---|---|
| r | leitura |
| w | escrita |
| a | adicionar |



# 💻 Criando arquivo


````python

arquivo = open("produtos.txt", "w")

arquivo.write("Notebook")

arquivo.close()
````

# 🧠 Parte 3 — Escrita de Arquivos

# 💻 Gravando múltiplas informações


````python

arquivo = open("clientes.txt", "w")

arquivo.write("Carlos\n")

arquivo.write("Maria\n")

arquivo.write("João\n")

arquivo.close()
````


# 🔍 O que é \n ?

Quebra de linha.



# 💻 Adicionando dados sem apagar conteúdo


````python

arquivo = open("clientes.txt", "a")

arquivo.write("Ana\n")

arquivo.close()
````


# 💻 Salvando produtos


````python

produto = input("Digite o produto: ")

arquivo = open("produtos.txt", "a")

arquivo.write(produto + "\n")

arquivo.close()

print("Produto salvo")
````


# 🧠 Parte 4 — Leitura de Arquivos

# 💻 Lendo arquivo completo


````python

arquivo = open("clientes.txt", "r")

conteudo = arquivo.read()

print(conteudo)

arquivo.close()
````


# 💻 Lendo linha por linha


````python

arquivo = open("clientes.txt", "r")

for linha in arquivo:

    print(linha)

arquivo.close()
````


# 💻 Armazenando linhas em lista


````python

arquivo = open("clientes.txt", "r")

linhas = arquivo.readlines()

print(linhas)

arquivo.close()
````


# 💻 Sistema de listagem


````python

arquivo = open("produtos.txt", "r")

for produto in arquivo:

    print(produto)

arquivo.close()
````


# 🧠 Parte 5 — Persistência em Sistemas

# 📖 O que é persistência?

Persistência é a capacidade de armazenar dados permanentemente.



# 💡 Sem persistência

❌ dados são perdidos ao fechar o programa.



# 💡 Com persistência

✅ dados permanecem salvos.



# 💻 Exemplo — Cadastro persistente


````python

def cadastrar():

    produto = input("Digite o produto: ")

    arquivo = open("produtos.txt", "a")

    arquivo.write(produto + "\n")

    arquivo.close()

    print("Produto salvo")

cadastrar()
````


# 💻 Exemplo — Relatório persistente


````python

def listar():

    arquivo = open("produtos.txt", "r")

    for produto in arquivo:

        print(produto)

    arquivo.close()

listar()
````


# 🛒 Parte 6 — Projeto do Semestre

# 📦 Sistema de Gestão Comercial

Hoje iremos adicionar persistência ao sistema.



# 💻 Sistema completo com arquivos


````python

def cadastrar_produto():

    produto = input("Digite o produto: ")

    arquivo = open("produtos.txt", "a")

    arquivo.write(produto + "\n")

    arquivo.close()

    print("Produto cadastrado")

def listar_produtos():

    arquivo = open("produtos.txt", "r")

    for produto in arquivo:

        print(produto)

    arquivo.close()

opcao = ""

while opcao != "0":

    print("1 - Cadastrar produto")
    print("2 - Listar produtos")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":

        cadastrar_produto()

    if opcao == "2":

        listar_produtos()
````
