# 🧠 Parte 2 — Introdução ao FOR

# 📖 O que é FOR?

FOR é uma estrutura de repetição utilizada quando sabemos quantas vezes o bloco será executado.



# 💻 Exemplo simples

```python
for numero in range(5):
    print(numero)
```


````python

for numero in range(5):
    print(numero)
````


# 🔍 O que aconteceu?

O Python executou o bloco:

✅ 5 vezes  
✅ iniciando em 0  
✅ terminando em 4  



# 📖 Sintaxe do FOR

```python
for variável in range():
    bloco de código
```



# 🧠 Parte 3 — Range e Intervalos

# 📖 O que é range()?

A função range() gera sequências numéricas.



# 💻 Exemplo — Range simples


````python

for numero in range(10):
    print(numero)
````


# 💻 Exemplo — Range com início e fim


````python

for numero in range(1, 6):
    print(numero)
````


# 💻 Exemplo — Incremento personalizado


````python

for numero in range(0, 11, 2):
    print(numero)
````


# 🔍 Explicação

```python
range(início, fim, incremento)
```



# 💻 Contagem regressiva


````python

for numero in range(10, 0, -1):
    print(numero)
````


# 🧠 Parte 4 — Iterações em Sistemas

# 💡 Sistemas utilizam FOR constantemente



# 💡 Exemplos reais

✅ geração de relatórios  
✅ processamento de listas  
✅ cadastro de produtos  
✅ sistemas financeiros  
✅ controle de estoque  



# 💻 Exemplo — Cadastro de produtos


````python

for contador in range(3):

    produto = input("Digite o produto: ")

    print("Produto cadastrado:", produto)
````


# 💻 Exemplo — Sistema de notas


````python

soma = 0

for contador in range(3):

    nota = float(input("Digite a nota: "))

    soma = soma + nota

media = soma / 3

print("Média:", media)
````


# 💻 Exemplo — Relatório simples


````python

for contador in range(5):

    print("Relatório gerado")
````


# 🧠 Parte 5 — Tabelas e Padrões

# 💻 Tabuada


````python

numero = int(input("Digite um número: "))

for contador in range(1, 11):

    resultado = numero * contador

    print(numero, "x", contador, "=", resultado)
````


# 💻 Padrão de caracteres


````python

for contador in range(5):

    print("*" * 5)
````


# 💻 Sequência numérica


````python

for numero in range(1, 21):

    print(numero)
````


# 🛒 Parte 6 — Projeto do Semestre
