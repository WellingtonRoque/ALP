# 🧠 Parte 2 — Operadores Relacionais

# 📖 O que são operadores relacionais?

São operadores utilizados para comparar valores.



# 📋 Principais operadores relacionais

| Operador | Significado |
|---|---|
| > | Maior |
| < | Menor |
| >= | Maior ou igual |
| <= | Menor ou igual |
| == | Igual |
| != | Diferente |



# 💻 Exemplo — Maior que


````python

numero = 10

print(numero > 5)
````


# 🔍 Resultado

```python
True
```



# 💻 Exemplo — Menor que


````python

idade = 15

print(idade < 18)
````


# 💻 Exemplo — Igualdade


````python

senha = 1234

print(senha == 1234)
````


# 💻 Exemplo — Diferença


````python

produto = "Mouse"

print(produto != "Teclado")
````


# 🧠 Parte 3 — Operadores Lógicos

# 📖 O que são operadores lógicos?

São utilizados para combinar comparações.



# 📋 Principais operadores lógicos

| Operador | Função |
|---|---|
| and | E |
| or | OU |
| not | NÃO |



# 💻 Operador AND

As duas condições precisam ser verdadeiras.


````python

idade = 20

print(idade >= 18 and idade <= 60)
````


# 💻 Operador OR

Apenas uma condição precisa ser verdadeira.


````python

cidade = "Indaiatuba"

print(cidade == "Indaiatuba" or cidade == "Campinas")
````


# 💻 Operador NOT

Inverte o resultado lógico.


````python

ativo = True

print(not ativo)
````


# 🧠 Parte 4 — Expressões Computacionais

# 📖 O que é uma expressão?

É uma combinação de operadores e valores.



# 💻 Exemplo


````python

idade = 25

resultado = idade >= 18

print(resultado)
````


# 💻 Expressões compostas


````python

idade = 25

salario = 3000

resultado = idade >= 18 and salario >= 2000

print(resultado)
````
