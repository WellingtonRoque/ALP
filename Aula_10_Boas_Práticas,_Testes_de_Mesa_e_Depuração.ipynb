
# 🧠 Parte 4 — Testes de Mesa

# 📖 O que é teste de mesa?

Técnica utilizada para acompanhar a execução do algoritmo manualmente.



# 💻 Exemplo


```python

contador = 1

while contador <= 3:

    print(contador)

    contador = contador + 1
```


# 🔍 Teste de Mesa

| contador | condição | saída |
|---|---|---|
| 1 | True | 1 |
| 2 | True | 2 |
| 3 | True | 3 |
| 4 | False | fim |



# 💻 Exemplo — Soma acumulativa


```python

contador = 1

soma = 0

while contador <= 5:

    soma = soma + contador

    contador = contador + 1

print(soma)
```


# 🔍 Teste de Mesa

| contador | soma |
|---|---|
| 1 | 1 |
| 2 | 3 |
| 3 | 6 |
| 4 | 10 |
| 5 | 15 |



# 🧠 Parte 5 — Depuração de Programas

# 📖 O que é depuração?

Processo de encontrar erros no código.



# 💡 Tipos de erros

✅ erro de sintaxe  
✅ erro lógico  
✅ erro de execução  



# ❌ Exemplo — Erro de sintaxe


# ❌ Exemplo — Erro lógico


```python

numero1 = 10

numero2 = 5

resultado = numero1 - numero2

print(resultado)
```


# ⚠️ O algoritmo deveria somar



# 💻 Utilizando print para depuração


```python

numero = 10

print("Valor atual:", numero)

numero = numero + 5

print("Novo valor:", numero)
```


# 💻 Depuração em loops


```python

contador = 1

while contador <= 3:

    print("Executando:", contador)

    contador = contador + 1
```
