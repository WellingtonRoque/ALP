# 🧠 Parte 2 — Métodos Avançados de Listas

# 📖 Métodos são funções próprias da lista.



# 💻 append()

Adiciona elementos na lista.


```python

produtos = []

produtos.append("Notebook")

produtos.append("Mouse")

print(produtos)
```


# 💻 remove()

Remove elementos da lista.


```python

produtos = ["Notebook", "Mouse", "Monitor"]

produtos.remove("Mouse")

print(produtos)
```


# 💻 pop()

Remove elemento utilizando índice.


```python

produtos = ["Notebook", "Mouse", "Monitor"]

produtos.pop(1)

print(produtos)
```


# 💻 clear()

Remove todos os elementos.


```python

produtos = ["Notebook", "Mouse"]

produtos.clear()

print(produtos)
```


# 💻 insert()

Adiciona elemento em posição específica.


```python

produtos = ["Notebook", "Monitor"]

produtos.insert(1, "Mouse")

print(produtos)
```


# 🧠 Parte 3 — Busca de Elementos

# 📖 Sistemas frequentemente precisam localizar dados.



# 💻 Verificando existência


```python

produtos = ["Notebook", "Mouse", "Monitor"]

print("Mouse" in produtos)
```


# 💻 Busca utilizando IF


```python

produtos = ["Notebook", "Mouse", "Monitor"]

produto = input("Digite o produto: ")

if produto in produtos:

    print("Produto encontrado")

else:

    print("Produto não encontrado")
```


# 💻 index()

Localiza posição do elemento.


```python

produtos = ["Notebook", "Mouse", "Monitor"]

print(produtos.index("Mouse"))
```


# ⚠️ Atenção

Se o elemento não existir:

❌ erro no sistema



# 💻 Busca segura


```python

produtos = ["Notebook", "Mouse", "Monitor"]

produto = input("Digite o produto: ")

if produto in produtos:

    indice = produtos.index(produto)

    print("Posição:", indice)

else:

    print("Produto não encontrado")
```


# 🧠 Parte 4 — Ordenação de Listas

# 📖 Sistemas utilizam ordenação constantemente.



# 💡 Exemplos reais

✅ produtos  
✅ clientes  
✅ relatórios  
✅ rankings  
✅ vendas  



# 💻 sort()

Ordena lista em ordem crescente.


```python

valores = [50, 10, 90, 20]

valores.sort()

print(valores)
```


# 💻 reverse()

Inverte lista.


```python

valores = [10, 20, 30]

valores.reverse()

print(valores)
```


# 💻 sorted()

Retorna nova lista ordenada.


```python

nomes = ["Carlos", "Ana", "Maria"]

resultado = sorted(nomes)

print(resultado)
```


# 💻 Ordenação decrescente


```python

valores = [10, 40, 20]

valores.sort(reverse=True)

print(valores)
```


# 🧠 Parte 5 — Manipulação Dinâmica

# 💻 Cadastro dinâmico de produtos


```python

produtos = []

opcao = ""

while opcao != "0":

    print("1 - Cadastrar")
    print("2 - Listar")
    print("3 - Remover")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":

        produto = input("Digite o produto: ")

        produtos.append(produto)

    if opcao == "2":

        print(produtos)

    if opcao == "3":

        produto = input("Digite o produto para remover: ")

        if produto in produtos:

            produtos.remove(produto)

            print("Produto removido")
```


# 💻 Sistema de ordenação


```python

produtos = ["Mouse", "Notebook", "Monitor"]

produtos.sort()

print(produtos)
```


# 🛒 Parte 6 — Projeto do Semestre

# 📦 Sistema de Gestão Comercial

Hoje iremos adicionar busca e ordenação ao sistema.



# 💻 Sistema completo de produtos


```python

produtos = []

opcao = ""

while opcao != "0":

    print("===== PRODUTOS =====")
    print("1 - Cadastrar")
    print("2 - Listar")
    print("3 - Buscar")
    print("4 - Ordenar")
    print("0 - Sair")

    opcao = input("Digite a opção: ")

    if opcao == "1":

        produto = input("Digite o produto: ")

        produtos.append(produto)

    if opcao == "2":

        print(produtos)

    if opcao == "3":

        busca = input("Digite o produto: ")

        if busca in produtos:

            print("Produto encontrado")

        else:

            print("Produto não encontrado")

    if opcao == "4":

        produtos.sort()

        print(produtos)
```
