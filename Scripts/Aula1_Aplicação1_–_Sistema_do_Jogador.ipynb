# 🟦 Parte 1 – Bloco que Muda de Cor


```

script.Parent.Touched:Connect(function(hit)
    local personagem = hit.Parent

    if personagem:FindFirstChild("Humanoid") then
        script.Parent.Color = Color3.new(0, 1, 0)
    end
end)
```


# 🎯 Atividade 1

Faça o bloco mudar para:

- azul
- vermelho
- amarelo



# 🟨 Parte 2 – Bloco com Mensagem

# **🎮 Hint**

Ele mostra mensagem no topo da tela.

```
script.Parent.Touched:Connect(function()
    local mensagem = Instance.new("Hint")

    mensagem.Text = "Você encontrou um bloco especial!"

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()
end)
```

🎯 O QUE ACONTECE

Quando o jogador toca:

# 🟩 Parte 3 – Bloco de Pontos


```

local pontos = 0

script.Parent.Touched:Connect(function()
    pontos = pontos + 10

    print(pontos)
end)
```


# 🎯 Atividade 3

# **🎮 VERSÃO EVOLUÍDA - OBJETIVO**

Quando o jogador tocar no bloco:

✔ ganhar pontos

✔ mostrar mensagem na tela

✔ mostrar quantos pontos ganhou

```
local pontos = 0

script.Parent.Touched:Connect(function()
    pontos = pontos + 10

    local mensagem = Instance.new("Hint")

    mensagem.Text = "Pontos: " .. pontos

    mensagem.Parent = game.Workspace

    wait(10)

    mensagem:Destroy()
end)
```


# 🟥 Parte 4 – Bloco de Dano


```

local vida = 100

script.Parent.Touched:Connect(function()
    vida = vida - 10

    print(vida)
end)
```


# 🎯 Atividade 4

Faça:

- perder 20 de vida
- depois perder mais 30
- mostrar resultado final


# **🟥 Evolução do Bloco de Dano – Sistema de Vida**

Quando o jogador tocar no bloco:

perderá vida
o bloco ficará vermelho
uma mensagem aparecerá na tela

```
local vida = 100

script.Parent.Touched:Connect(function()

    vida = vida - 10

    script.Parent.Color = Color3.new(1, 0, 0)

    local mensagem = Instance.new("Hint")

    mensagem.Text = "Vida: " .. vida

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()

end)
```


# 🟪 Parte 5 – Bloco de Energia


```

local energia = 50

script.Parent.Touched:Connect(function()
    energia = energia + 20

    print(energia)
end)
```


# 🎯 Atividade 5

# **🟩 Bloco de Energia – Power-up de Energia**

Quando o jogador tocar no bloco:

ganhará energia

o bloco ficará verde

uma mensagem aparecerá na tela

```
local energia = 50

script.Parent.Touched:Connect(function()

    energia = energia + 20

    script.Parent.Color = Color3.new(0, 1, 0)

    local mensagem = Instance.new("Hint")

    mensagem.Text = "Energia: " .. energia

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()

end)
```
