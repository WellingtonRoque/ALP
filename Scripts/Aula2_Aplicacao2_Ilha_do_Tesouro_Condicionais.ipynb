
# 🎮 Aplicando no Roblox

Agora vamos transformar essas regras em objetos reais do jogo.



# 🏆 Baú do Tesouro


````

local moedas = 50

script.Parent.Touched:Connect(function()

    local mensagem = Instance.new("Hint")

    if moedas >= 50 then
        mensagem.Text = "🏆 Baú Aberto!"
    else
        mensagem.Text = "Você precisa de mais moedas."
    end

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()

end)
````


# 🚪 Portal Secreto


````

local chave = true

script.Parent.Touched:Connect(function()

    local mensagem = Instance.new("Hint")

    if chave then
        mensagem.Text = "🚪 Portal Liberado"
    else
        mensagem.Text = "🔒 Portal Bloqueado"
    end

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()

end)
````


# ⚡ Cristal de Energia


````

local energia = 90

script.Parent.Touched:Connect(function()

    energia = energia + 20

    local mensagem = Instance.new("Hint")

    if energia >= 100 then
        mensagem.Text = "⚡ Energia Máxima"
    else
        mensagem.Text = "Energia: " .. energia
    end

    mensagem.Parent = game.Workspace

    wait(3)

    mensagem:Destroy()

end)
````
