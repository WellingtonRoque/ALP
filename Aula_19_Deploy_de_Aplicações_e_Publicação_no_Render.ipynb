
# 🧠 Parte 4 — Publicação no Render

# 📖 O que é Render?

Plataforma gratuita de hospedagem de aplicações.



# 💡 Passos para deploy

1. criar conta no Render  
2. conectar GitHub  
3. selecionar repositório  
4. configurar aplicação  
5. publicar sistema  



# 💻 Configuração principal


````python

Build Command:

pip install -r requirements.txt
````


# 💻 Start Command


````python

python app.py
````


# 💻 Deploy automático

Sempre que houver push no GitHub:

✅ Render atualiza automaticamente o sistema.
